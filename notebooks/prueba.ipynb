```python
import pandas as pd

ventas = pd.DataFrame({
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam'],
    'categoria': ['Electrónica', 'Periféricos', 'Periféricos', 'Electrónica', 'Periféricos'],
    'precio': [1200, 25, 45, 350, 60],
    'unidades_vendidas': [150, 2000, 800, 300, 500]
})

# Ver las primeras filas
ventas.head()

#print(ventas)
print(f"\nForma: {ventas.shape}")  # (filas, columnas)
print(f"Columnas: {list(ventas.columns)}")

# SELECCIONAR una columna (como SELECT columna FROM tabla)
print(ventas['producto'])

# FILTRAR filas (como WHERE)
caros = ventas[ventas['precio'] > 100]
print(caros)

# AGREGAR (como GROUP BY + SUM)
por_categoria = ventas.groupby('categoria')['precio'].mean()
print(por_categoria)

# Crear columna calculada
ventas['ingresos'] = ventas['precio'] * ventas['unidades_vendidas']
print(ventas[['producto', 'ingresos']].sort_values('ingresos', ascending=False))
```
