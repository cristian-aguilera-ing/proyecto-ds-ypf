# Leyendo el dataset

```python
import pandas as pd

df = pd.read_csv(
    '../data/raw/produccin-de-pozos-de-gas-y-petrleo-no-convencional.csv',
    sep=',',
    encoding='utf-8-sig',
    low_memory=False,
    parse_dates=['fechaingreso', 'fecha_data'],
    dtype={
        'anio': 'int16',
        'mes': 'int8',
        'idempresa': 'category',
        'empresa': 'category',
        'tipoextraccion': 'category',
        'tipoestado': 'category',
        'tipopozo': 'category',
        'cuenca': 'category',
        'provincia': 'category',
        'tipo_de_recurso': 'category',
    }
)

print(df.shape)
print(df.dtypes)
df.head()
```
