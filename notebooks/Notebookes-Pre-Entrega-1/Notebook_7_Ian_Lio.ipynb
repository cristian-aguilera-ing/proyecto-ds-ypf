14) ¿Cual es la expectativa de vida mas alta y mas baja? Usa la funcion de pandas y la funcion de NumPy y compara el resultado. ¿Que paises tienen esas expectativas de vida?

```python
# Maxima expectativa de vida usando numpy
#maxima = np.array(expvida['life_expectancy'])
#maxim_exp = np.max(array)
expvida[['life_expectancy']].describe()     #Nos da una descripcion general de nuestros datos, particularizamos a la primera columna
```

```python
# Maxima expectativa de vida usando Numpy
maxima = np.array(expvida['life_expectancy'])
maxim_exp = np.nanmax(maxima)      #Agregamos nan
                                   #Ojo que en Numpy los calculos son estrictos y toma los nulos, en contraposicion Pandas si los ignora
print(maxim_exp)
```

```python
# Maxima expectativa de vida usando pandas
maxima_exp = expvida['life_expectancy'].max()
print(maxima_exp)
```

```python
# Minima expectativa de vida usando numpy
minima = np.array(expvida['life_expectancy'])
minim_exp = np.nanmin(minima)
print(minim_exp)
```

```python
# Minima expectativa de vida usando pandas
minima_exp = expvida['life_expectancy'].min()
print(minim_exp)
```

```python
#Mostramos cuales son estos paises que disponen del maximo (minimo) expectativa de vida.

paises_maxima = expvida[expvida['life_expectancy'] == maxima_exp]['Country'].unique()
paises_minima = expvida[expvida['life_expectancy'] == minim_exp]['Country'].unique()

print(f"\nPaíses con la expectativa de vida más ALTA ({maxima_exp}): {paises_maxima}")
print(f"Países con la expectativa de vida más BAJA ({minim_exp}): {paises_minima}")
```

16) ¿Cuantos paises se ubican en cada una de esas categorias de "Status"? _**Pista**_: Lee acerca de la función [value_counts()](http://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.value_counts.html)





```python
#Contamos cuántas filas (registros de todos los años) hay por categoría
conteo_filas = expvida['Status'].value_counts()
print("Cantidad total de registros (filas) por categoría:")
print(conteo_filas)

#Tenemos que tener cuidado porque varias filas tienen el mismo pais
#Cuenta cuántos PAÍSES únicos hay por categoría
conteo_paises = expvida[['Country', 'Status']].drop_duplicates()['Status'].value_counts()
print("\nCantidad real de PAÍSES únicos por categoría:")
print(conteo_paises)
```

18) ¿Cuantos paises presentaron algun caso de Sarampion (**Measles**) en el 2015? **_Pista_**: Lee acerca de la función [.count()](http://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.count.html?highlight=count#pandas.DataFrame.count)

```python
#Filtramos el DataFrame con las dos condiciones (Año 2015 Y casos mayores a 0)
#Usamos el símbolo '&' para indicar que se tienen que cumplir ambas cosas
con_sarampion_2015 = expvida[(expvida['Year'] == 2015) & (expvida['measles'] > 0)]

# 2. Seleccionamos la columna de países y usamos .count() para contar las filas resultantes
cantidad_paises = con_sarampion_2015['Country'].count()

print(f"En el año 2015, {cantidad_paises} países presentaron al menos un caso de sarampión.")
```
