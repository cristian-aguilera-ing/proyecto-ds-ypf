```python
#Ejercicio 4: Crear una arreglo 2d de shape (3, 3) que tenga los números del 0 al 8
consecutivos=np.arange(0,9).reshape(3,3) #notar que .reshape(n,m) nos redimensiona el arreglo a nuestra voluntad. Se ve sumamente util.
                                        #Tambien se puede hacer luego de haber definido el arreglo con .shape: vector.shape=(n,m) siempre que n+m de la suma de la misma cantidad de elemtos del orginial
print(consecutivos)
```

```python
#EJERCICIO ADICIONAL:
#Toma el ejercicio anterior y usando el la funcion reshape, haz que sea un vector nuevamente
consecutivos=np.arange(0,9).reshape(3,3)
consecutivos=consecutivos.reshape(9,)     #Notar que cuando dejamos vacio la cordenada 2: .rshape(n+m, ) significa que lo volvemos la matriz a un vector (una sola x-upla)
print(consecutivos)
```

```python
#Ejercicio 5: Crear un arreglo de números aleatorios de longitud 100 y obtener su media y varianza
vecaz=np.random.rand(100)       #.random.rand(n) solo nos permite que los n numeros aleatorios pertenezcan al intervalo [0,1]


prom=np.mean(vecaz)
var=np.var(vecaz)


print(f"Los numeros aleatorios son {vecaz} El promedio es {prom} y la varianza es {var}")

```

```python
#Ejercicio 6: Calcular la media de un arreglo usando np.sum y la funcion len
vector = np.array([1,2,3,4,5,-6])
prom = (np.sum(vector) )/len(vector)  #np.sum(arreglo) suma todas los elementos del vector
print(prom)
```

```python
#ejercicio 7: Calcular la varianza de un arreglo usando np.sum y np.mean
vector = np.array([1,2,3,4,5])
var = (np.sum((vector-np.mean(vector))**2))/len(vector)
print(var)
```
